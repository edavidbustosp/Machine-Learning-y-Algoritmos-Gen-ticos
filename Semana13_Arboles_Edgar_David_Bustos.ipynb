## 2. Demostración guiada (resuelta)
Construimos un árbol de una universidad. Observe que cada nodo tiene una **etiqueta** y un **identificador único**.

```python
universidad = Tree()
universidad.create_node("Universidad", "u")
universidad.create_node("Facultad de Ingeniería", "fi", parent="u")
universidad.create_node("Facultad de Educación", "fe", parent="u")
universidad.create_node("Ingeniería de Sistemas", "sis", parent="fi")
universidad.create_node("Ingeniería Industrial", "ind", parent="fi")
universidad.show()
print("Padre de Sistemas:", universidad.parent("sis").tag)
print("Hijos de Ingeniería:", [n.tag for n in universidad.children("fi")])
```

### Operaciones adicionales demostradas

```python
universidad.update_node("ind", tag="Ingeniería Industrial y Logística")
print("Nodos en profundidad:", [universidad.get_node(i).tag for i in universidad.expand_tree(mode=Tree.DEPTH)])
print("Nodos por niveles:", [universidad.get_node(i).tag for i in universidad.expand_tree(mode=Tree.WIDTH)])
universidad.show()
```

```python
# TODO: implementar el ejercicio 1
catalogo = Tree()
catalogo.create_node("Tienda", "tie")
catalogo.create_node("Tecnología", "tec", parent="tie")
catalogo.create_node("Hogar", "hog", parent="tie")
catalogo.create_node("Libros", "lib", parent="tie")
catalogo.create_node("Impresoras", "prin", parent="tec")
catalogo.create_node("Monitores", "monit", parent="tec")
catalogo.create_node("Jabón Rey","rey", parent="hog")
catalogo.create_node("Binipel", "bini", parent="hog")
catalogo.create_node("El Señor de los Anillos","ring",parent="lib")
catalogo.create_node("La Culpa es de la Vaca", "vaca", parent="lib")

catalogo.show()
```
